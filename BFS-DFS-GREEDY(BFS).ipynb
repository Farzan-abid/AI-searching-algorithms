```python
from typing import OrderedDict
##apply bfs
from collections import deque
def bfs(graph,start,goal):
  queue=deque()
  order=deque()
  visited=set()
  node_expanded=0
  queue.append(start)
  while queue:
    node=queue.popleft()
    order.append(node)
    # if node==goal:
    #   queue.append(node)
    #   return order,node_expanded
    node_expanded+=1
    for successor in graph[node]:
      if successor not in visited:
        if successor==goal:
           queue.append(node)
           return order,node_expanded
        else:
         queue.append(successor)
         visited.add(successor)






```

```python
print(bfs(graph,'S','G'))
```

```python
def dfs(graph,start,goal):
  stack=deque()
  order=[start]
  stack.append(start)
  visited=set()

  while stack:
    node=stack.pop()
    if node==goal:
      stack.pop()
      visited.add(node)
    if node not in visited:
      visited.add(node)
      order.append(node)
      for neighbor in graph[node]:
            if neighbor not in visited:
             stack.append(neighbor)
  return order,stack


```

```python
print(dfs(graph,'S','G'))
```
