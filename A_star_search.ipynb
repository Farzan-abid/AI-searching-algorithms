```python
graph = {
    "A": {"B": 1, "C": 4},
    "B": {"A": 1, "D": 2, "E": 5},
    "C": {"A": 4, "D": 1},
    "D": {"B": 2, "C": 1, "E": 1},
    "E": {"B": 5, "D": 1}
}
heuristics = {
    "A": 6,
    "B": 4,
    "C": 4,
    "D": 1,
    "E": 0
}
```

```python
def path_find(came_from,goal):
  current=goal
  path=[]
  path.append(goal)
  while came_from[current] != None:


    current=came_from[current]
    path.append(current)
  path.reverse()

  return  path
def get_cost(graph,path):
  cost=0
  for i in range(len(path)-1):
    cost+=graph[path[i]][path[i+1]]
  return cost
```

```python
import heapq

def a_star_search(graph, heuristics, start, goal):
    frontier = [(heuristics[start], start)]    # (f(n), node), f(start) = 0 + h(start)
    g_cost = {start: 0}                         # cheapest known cost to reach each node
    came_from = {start: None}

    while frontier:
        # TODO 1: pop the node with the lowest f(n) from the frontier
        p,current=heapq.heappop(frontier)
        #         and store only the node name in 'current'


        if current == goal:
            break

        for neighbor, cost in graph[current].items():
            # TODO 2: compute new_g, the cost of reaching 'neighbor' through 'current'
            new_g=g_cost[current]+graph[current][neighbor]


            # TODO 3: if 'neighbor' is new OR new_g is cheaper than g_cost[neighbor]:
            if neighbor not in g_cost or new_g<g_cost[neighbor]:
            #           a) store new_g in g_cost[neighbor]
             g_cost[neighbor]=new_g
            #           b) compute f = new_g + heuristics[neighbor]
             f = new_g + heuristics[neighbor]
            #           c) push (f, neighbor) onto the frontier
             heapq.heappush(frontier,(f,neighbor))

            #           d) set came_from[neighbor] = current
             came_from[neighbor]=current
            pass

    return path_find(came_from, goal), get_cost(graph,path_find(came_from, goal))
```

```python
path, total_cost = a_star_search(graph, heuristics, "A", "E")
print("A* path:", path)
print("Total cost:", total_cost)
```
